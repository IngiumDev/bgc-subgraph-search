```python
# So know we want per run a list of the taxa present
run_taxes = {

}
for run_id, graph_to_tax in run_graph_to_tax.items():
    taxa_in_run = []
    for graph_id, tax_summary in graph_to_tax.items():
        taxa_in_run.append(tax_summary)
    run_taxes[run_id] = taxa_in_run
```
